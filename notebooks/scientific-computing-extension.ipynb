## Regression-Style Experimentation

The reviewed notebook included regression experiments with `LinearRegression` and `Ridge`. The cell below keeps the idea but uses aggregate item-level data only. It is a technical demonstration, not a participant-level predictive model and not an official ECHO-UFS result.

````python
features = instructional_items[["std", "kruskal_p"]]
target = instructional_items["mean"]

models = {
    "LinearRegression": LinearRegression(),
    "Ridge": Ridge(alpha=1.0),
}

model_scores = []
for name, model in models.items():
    model.fit(features, target)
    predictions = model.predict(features)
    model_scores.append({
        "model": name,
        "r2_on_aggregate_items": r2_score(target, predictions),
        "intercept": model.intercept_,
    })

pd.DataFrame(model_scores)
````

````python
x = instructional_items["item_id"].to_numpy().reshape(-1, 1)
y = instructional_items["mean"].to_numpy()
trend_model = LinearRegression().fit(x, y)
trend = trend_model.predict(x)

plt.figure(figsize=(8, 4))
plt.scatter(instructional_items["item_id"], y, label="Aggregate item mean")
plt.plot(instructional_items["item_id"], trend, color="#8b4d57", label="Linear trend")
plt.xlabel("Evaluation item index")
plt.ylabel("Mean score")
plt.title("Aggregate Item-Level Trend Demonstration")
plt.legend()
plt.tight_layout()
plt.show()

print({
    "slope": float(trend_model.coef_[0]),
    "intercept": float(trend_model.intercept_),
    "r2": float(r2_score(y, trend)),
})
````
