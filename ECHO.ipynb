```python
# Calcular frequências absolutas
freq_absoluta <- dataset %>%
  gather(key = "Pergunta", value = "Resposta", -`Escolha seu Estado`) %>%
  group_by(`Escolha seu Estado`, Pergunta, Resposta) %>%
  summarise(Frequencia = n()) %>%
  arrange(desc(Frequencia))

# Calcular frequências relativas
freq_relativa <- freq_absoluta %>%
  group_by(Pergunta) %>%
  mutate(Frequencia_Relativa = Frequencia / sum(Frequencia))

# Visualizar frequências absolutas e relativas
print(freq_absoluta)
print(freq_relativa)
```

```python
# Calcular mediana e quartis para cada pergunta
estatisticas <- dataset %>%
  gather(key = "Pergunta", value = "Resposta", -`Escolha seu Estado`) %>%
  group_by(Pergunta) %>%
  summarise(
    Mediana = median(as.numeric(Resposta), na.rm = TRUE),
    Q1 = quantile(as.numeric(Resposta), 0.25, na.rm = TRUE),
    Q3 = quantile(as.numeric(Resposta), 0.75, na.rm = TRUE)
  )

# Visualizar resultados
print(estatisticas)
```

```python
# Função para calcular moda
moda <- function(v) {
  uniqv <- unique(v)
  uniqv[which.max(tabulate(match(v, uniqv)))]
}

# Calcular a moda para cada pergunta
moda_resposta <- dataset %>%
  gather(key = "Pergunta", value = "Resposta", -`Escolha seu Estado`) %>%
  group_by(Pergunta) %>%
  summarise(Moda = moda(Resposta))

# Visualizar a moda
print(moda_resposta)
```

```python
# Primeiro, instalar bibliotecas se precisar
!pip install openpyxl scikit-learn seaborn

# Importar bibliotecas
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# ==========================
# Análise dos Dados Carregados
# ==========================

# Visualizar as primeiras linhas (opcional)
print("\nVisualização Inicial do DataFrame:")
print(df.head())

# Mostrar todas as colunas disponíveis
print("\nColunas disponíveis no DataFrame:")
print(df.columns.tolist())

# ==========================
# Seleção das colunas válidas para análise
# ==========================

# Vamos excluir manualmente colunas que não são respostas
colunas_excluir = ['Escolha seu Estado', 'Satisfacao_Curso']  # Ajuste aqui se precisar
colunas_uteis = [col for col in df.columns if col not in colunas_excluir and df[col].dtype != 'O']

# Subconjunto do dataframe apenas com colunas numéricas
df_perguntas = df[colunas_uteis].copy()

print("\nColunas utilizadas para análise:")
print(df_perguntas.columns.tolist())

# ==========================
# 1. Análise Descritiva
# ==========================

print("\nTabela Descritiva das Perguntas:")
print(df_perguntas.describe().T)

# ==========================
# 2. Teste de Confiabilidade - Alfa de Cronbach
# ==========================

def cronbach_alpha(df_subset):
    df_corr = df_subset.corr()
    n_items = len(df_corr)
    mean_corr = df_corr.values[np.triu_indices(n_items, 1)].mean()
    alpha = (n_items * mean_corr) / (1 + (n_items - 1) * mean_corr)
    return alpha

if not df_perguntas.empty:
    alpha = cronbach_alpha(df_perguntas)
    print(f"\nAlfa de Cronbach da Escala: {alpha:.3f}")
else:
    print("\n⚠️ DataFrame de perguntas vazio. Alfa de Cronbach não pode ser calculado.")

# ==========================
# 3. Teste de Normalidade - Kolmogorov-Smirnov
# ==========================

print("\nTeste de Normalidade (Kolmogorov-Smirnov) para cada pergunta:")
for col in df_perguntas.columns:
    stat, p_value = stats.kstest(df_perguntas[col].dropna(), 'norm', args=(df_perguntas[col].mean(), df_perguntas[col].std()))
    print(f"{col}: KS = {stat:.4f}, p-value = {p_value:.4f}")

# ==========================
# 4. Teste t para Amostras Dependentes (Dois Momentos)
# ==========================

# Procurar colunas de momentos
moment1_cols = [col for col in df.columns if '_Momento1' in col]
moment2_cols = [col for col in df.columns if '_Momento2' in col]

print("\nTeste t para amostras relacionadas (Momentos 1 vs 2):")
if moment1_cols and moment2_cols:
    for m1, m2 in zip(moment1_cols, moment2_cols):
        if m1 in df.columns and m2 in df.columns:
            t_stat, p_value = stats.ttest_rel(df[m1].dropna(), df[m2].dropna())
            print(f"{m1} vs {m2}: t = {t_stat:.4f}, p-value = {p_value:.4f}")
else:
    print("⚠️ Não foram encontradas colunas dos dois momentos no DataFrame.")

# ==========================
# 5. Correlação de Spearman
# ==========================

print("\nCorrelação de Spearman entre as perguntas:")
if not df_perguntas.empty:
    corr_spearman = df_perguntas.corr(method='spearman')
    print(corr_spearman)

    # Plotar a matriz de correlação
    plt.figure(figsize=(12, 10))
    sns.heatmap(corr_spearman, annot=True, cmap='coolwarm', fmt=".2f")
    plt.title('Matriz de Correlação de Spearman')
    plt.show()
else:
    print("⚠️ DataFrame de perguntas vazio. Não é possível calcular correlação.")

# ==========================
# 6. Regressão Linear para prever Satisfação
# ==========================

print("\nRegressão Linear para prever Satisfação com o Curso:")

target = 'Satisfacao_Curso'

if target in df.columns:
    X = df_perguntas.copy()
    y = df[target]

    if not X.empty:
        model = LinearRegression()
        model.fit(X, y)
        y_pred = model.predict(X)

        print(f"R²: {r2_score(y, y_pred):.3f}")
        print(f"Intercepto: {model.intercept_}")

        coef_table = pd.DataFrame({
            'Variável': X.columns,
            'Coeficiente': model.coef_
        })
        print(coef_table)

        # Plotar comparação real vs predito
        plt.figure(figsize=(8,6))
        plt.scatter(y, y_pred)
        plt.xlabel('Satisfação Real')
        plt.ylabel('Satisfação Predita')
        plt.title('Satisfação Real vs Predita')
        plt.plot([y.min(), y.max()], [y.min(), y.max()], 'k--')
        plt.show()
    else:
        print("⚠️ Nenhum preditor disponível para regressão.")
else:
    print("⚠️ A coluna 'Satisfacao_Curso' não foi encontrada no DataFrame.")
```

```python
from scipy import stats

print("\n🧪 Teste Qui-Quadrado para cada pergunta (coluna):")

for col in df.columns[1:]:  # Ignora a primeira coluna se for identificadora
    # Remove valores ausentes (NaN) para não afetar o teste
    counts = df[col].value_counts().sort_index()

    if len(counts) < 2:
        print(f"{col}: ❌ Teste não aplicável (menos de 2 categorias)")
        continue

    chi2, p = stats.chisquare(counts)
    print(f"{col}: χ² = {chi2:.2f}, p-valor = {p:.4f}")
```

```python
from scipy import stats

agrupador = 'Escolha seu Estado'  # Altere aqui para outra variável, se necessário

print("\n📊 Teste de Kruskal-Wallis para cada pergunta por grupo de:", agrupador)

for col in df.columns[1:]:
    if col == agrupador:
        continue  # Evita comparar a coluna de agrupamento com ela mesma

    # Agrupar as respostas por grupo (ex: Estado)
    groups = [df[col][df[agrupador] == group].dropna() for group in df[agrupador].dropna().unique()]

    # Verifica se há pelo menos dois grupos com dados válidos
    if len(groups) < 2 or any(len(g) == 0 for g in groups):
        print(f"{col}: ❌ Teste não aplicável (menos de 2 grupos com dados)")
        continue

    try:
        kruskal_stat, p_value = stats.kruskal(*groups)
        print(f"{col}: H = {kruskal_stat:.2f}, p-valor = {p_value:.4f}")
    except ValueError as e:
        print(f"{col}: ⚠️ Erro ao executar o teste: {e}")
```

```python
# ==========================
# Instalar bibliotecas (se necessário no Colab)
# ==========================
!pip install openpyxl scikit-learn seaborn

# ==========================
# Importar bibliotecas
# ==========================
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# ==========================
# Visualização inicial do DataFrame
# ==========================
print("\n📌 Visualização Inicial do DataFrame:")
print(df.head())

print("\n📌 Colunas disponíveis no DataFrame:")
print(df.columns.tolist())

# ==========================
# Seleção de colunas úteis
# ==========================
colunas_excluir = ['Escolha seu Estado', 'Satisfacao_Curso']  # Ajustar se necessário
colunas_uteis = [col for col in df.columns if col not in colunas_excluir and df[col].dtype != 'O']

# Subconjunto apenas com colunas numéricas
df_perguntas = df[colunas_uteis].copy()

print("\n✅ Colunas utilizadas para análise:")
print(df_perguntas.columns.tolist())

# ==========================
# 1. Análise Descritiva
# ==========================
print("\n📊 Tabela Descritiva das Perguntas:")
print(df_perguntas.describe().T)

# ==========================
# 2. Alfa de Cronbach
# ==========================
def cronbach_alpha(df_subset):
    df_corr = df_subset.corr()
    n_items = len(df_corr)
    mean_corr = df_corr.values[np.triu_indices(n_items, 1)].mean()
    alpha = (n_items * mean_corr) / (1 + (n_items - 1) * mean_corr)
    return alpha

if not df_perguntas.empty:
    alpha = cronbach_alpha(df_perguntas)
    print(f"\n🧪 Alfa de Cronbach da Escala: {alpha:.3f}")
else:
    print("\n⚠️ DataFrame de perguntas vazio. Alfa de Cronbach não pode ser calculado.")

# ==========================
# 3. Teste de Normalidade (Kolmogorov-Smirnov)
# ==========================
print("\n📈 Teste de Normalidade (Kolmogorov-Smirnov):")
for col in df_perguntas.columns:
    stat, p_value = stats.kstest(df_perguntas[col].dropna(), 'norm', args=(df_perguntas[col].mean(), df_perguntas[col].std()))
    print(f"{col}: KS = {stat:.4f}, p-value = {p_value:.4f}")

# ==========================
# 4. Teste t para Momentos 1 e 2
# ==========================
moment1_cols = [col for col in df.columns if '_Momento1' in col]
moment2_cols = [col for col in df.columns if '_Momento2' in col]

print("\n🔁 Teste t para amostras relacionadas (Momentos 1 vs 2):")
if moment1_cols and moment2_cols:
    for m1, m2 in zip(moment1_cols, moment2_cols):
        if m1 in df.columns and m2 in df.columns:
            t_stat, p_value = stats.ttest_rel(df[m1].dropna(), df[m2].dropna())
            print(f"{m1} vs {m2}: t = {t_stat:.4f}, p-value = {p_value:.4f}")
else:
    print("⚠️ Não foram encontradas colunas dos dois momentos no DataFrame.")

# ==========================
# 5. Correlação de Spearman
# ==========================
print("\n🔗 Correlação de Spearman entre perguntas:")
if not df_perguntas.empty:
    corr_spearman = df_perguntas.corr(method='spearman')
    print(corr_spearman)

    # Gráfico da matriz de correlação
    plt.figure(figsize=(12, 10))
    sns.heatmap(corr_spearman, annot=True, cmap='coolwarm', fmt=".2f")
    plt.title('Matriz de Correlação de Spearman')
    plt.show()
else:
    print("⚠️ DataFrame de perguntas vazio. Não é possível calcular correlação.")

# ==========================
# 6. Regressão Linear
# ==========================
print("\n📉 Regressão Linear para prever 'Satisfacao_Curso':")
target = 'Satisfacao_Curso'

if target in df.columns:
    X = df_perguntas.copy()
    y = df[target]

    if not X.empty:
        model = LinearRegression()
        model.fit(X, y)
        y_pred = model.predict(X)

        print(f"R²: {r2_score(y, y_pred):.3f}")
        print(f"Intercepto: {model.intercept_:.3f}")

        coef_table = pd.DataFrame({
            'Variável': X.columns,
            'Coeficiente': model.coef_
        })
        print("\n📋 Coeficientes da Regressão:")
        print(coef_table)

        # Gráfico de predição
        plt.figure(figsize=(8,6))
        plt.scatter(y, y_pred)
        plt.xlabel('Satisfação Real')
        plt.ylabel('Satisfação Predita')
        plt.title('Satisfação Real vs Predita')
        plt.plot([y.min(), y.max()], [y.min(), y.max()], 'k--')
        plt.tight_layout()
        plt.show()
    else:
        print("⚠️ Nenhum preditor disponível para regressão.")
else:
    print("⚠️ A coluna 'Satisfacao_Curso' não foi encontrada no DataFrame.")
```

```python
# ==========================
# 📦 Instalar bibliotecas (Colab)
# ==========================
!pip install openpyxl scikit-learn seaborn

# ==========================
# 📚 Importar bibliotecas
# ==========================
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import IPython

# ==========================
# 📁 Leitura do arquivo Excel
# ==========================
file_path = '/content/gestaoConhecimentopt1.xlsx'  # Ajuste se necessário
df = pd.read_excel(file_path, header=1)
df.dropna(how='all', inplace=True)
df.dropna(axis=1, how='all', inplace=True)

print("\n# 📌 Visualização Inicial do DataFrame:")
print(df.head())
print("\n# 📌 Colunas disponíveis:")
print(df.columns.tolist())

# ==========================
# 📊 Frequência Absoluta, Relativa e Moda
# ==========================
print("\n## 📊 Frequência Absoluta, Relativa (%) e Moda para cada pergunta:")
for col in df.columns[1:]:
    print(f"\n🔸 Coluna: {col}")
    freq_abs = df[col].value_counts(dropna=False).sort_index()
    print("Frequência Absoluta:")
    print(freq_abs)
    freq_rel = df[col].value_counts(normalize=True, dropna=False).sort_index() * 100
    print("Frequência Relativa (%):")
    print(freq_rel.round(2))
    moda_val = df[col].mode(dropna=True)
    print("Moda:")
    print(moda_val.to_list())

# ==========================
# 🔍 Selecionar colunas numéricas
# ==========================
colunas_excluir = ['Escolha seu Estado', 'Satisfacao_Curso']
colunas_uteis = [col for col in df.columns if col not in colunas_excluir and df[col].dtype != 'O']
df_perguntas = df[colunas_uteis].copy()

print("\n# ✅ Colunas utilizadas para análise:")
print(df_perguntas.columns.tolist())

# ==========================
# 📊 RESUMO ESTATÍSTICO POR VARIÁVEL EM TABELA ÚNICA
# ==========================
agrupador = 'Escolha seu Estado'
print("\n## 📊 Tabela com Moda, Médias, Desvios, Testes Qui², Kruskal-Wallis e Normalidade:")
resultados = []

for col in df_perguntas.columns:
    dados_col = df[col].dropna()
    moda_val = df[col].mode(dropna=True)
    moda = moda_val.iloc[0] if not moda_val.empty else np.nan
    media = dados_col.mean()
    desvio = dados_col.std()
    try:
        freq = df[col].value_counts().sort_index()
        chi2_stat, chi2_p = stats.chisquare(freq)
    except:
        chi2_stat, chi2_p = np.nan, np.nan
    try:
        grupos = [df[col][df[agrupador] == g].dropna() for g in df[agrupador].dropna().unique()]
        if all(len(g) > 0 for g in grupos) and len(grupos) > 1:
            h_stat, h_p = stats.kruskal(*grupos)
        else:
            h_stat, h_p = np.nan, np.nan
    except:
        h_stat, h_p = np.nan, np.nan
    try:
        ks_stat, ks_p = stats.kstest(dados_col, 'norm', args=(media, desvio))
    except:
        ks_stat, ks_p = np.nan, np.nan
    resultados.append({
        'Variável': col,
        'Moda': moda,
        'Média': round(media, 4),
        'Desvio Padrão': round(desvio, 4),
        'Qui-Quadrado (Chi²)': round(chi2_stat, 4),
        'p-value Qui²': float(f"{chi2_p:.4e}") if not np.isnan(chi2_p) else np.nan,
        'Kruskal-Wallis H': round(h_stat, 4),
        'p-value Kruskal': round(h_p, 4) if not np.isnan(h_p) else np.nan,
        'KS': round(ks_stat, 4),
        'p-valor KS': round(ks_p, 4) if not np.isnan(ks_p) else np.nan
    })

df_resultados = pd.DataFrame(resultados)
IPython.display.display(df_resultados)

# ==========================
# 📊 1. Análise Descritiva
# ==========================
print("\n## 📊 Análise Descritiva")
print(df_perguntas.describe().T)

# ==========================
# 🧪 2. Alfa de Cronbach
# ==========================
def cronbach_alpha(df_subset):
    df_corr = df_subset.corr()
    n_items = len(df_corr)
    mean_corr = df_corr.values[np.triu_indices(n_items, 1)].mean()
    alpha = (n_items * mean_corr) / (1 + (n_items - 1) * mean_corr)
    return alpha

if not df_perguntas.empty:
    alpha = cronbach_alpha(df_perguntas)
    print(f"\n## 🧪 Alfa de Cronbach da Escala: {alpha:.3f}")

# ==========================
# 📈 3. Teste de Normalidade
# ==========================
print("\n## 📈 Teste de Normalidade (Kolmogorov-Smirnov)")
for col in df_perguntas.columns:
    stat, p_value = stats.kstest(df_perguntas[col].dropna(), 'norm', args=(df_perguntas[col].mean(), df_perguntas[col].std()))
    print(f"{col}: KS = {stat:.4f}, p-value = {p_value:.4f}")

# ==========================
# 🔁 4. Teste t para Momentos 1 e 2
# ==========================
moment1_cols = [col for col in df.columns if '_Momento1' in col]
moment2_cols = [col for col in df.columns if '_Momento2' in col]

print("\n## 🔁 Teste t para Momentos 1 vs 2")
if moment1_cols and moment2_cols:
    for m1, m2 in zip(moment1_cols, moment2_cols):
        if m1 in df.columns and m2 in df.columns:
            t_stat, p_value = stats.ttest_rel(df[m1].dropna(), df[m2].dropna())
            print(f"{m1} vs {m2}: t = {t_stat:.4f}, p-value = {p_value:.4f}")
else:
    print("⚠️ Colunas de momentos não encontradas.")

# ==========================
# 🔗 5. Correlação de Spearman
# ==========================
print("\n## 🔗 Correlação de Spearman")
if not df_perguntas.empty:
    corr_spearman = df_perguntas.corr(method='spearman')
    print(corr_spearman)
    plt.figure(figsize=(12, 10))
    sns.heatmap(corr_spearman, annot=True, cmap='coolwarm', fmt=".2f")
    plt.title('Matriz de Correlação de Spearman')
    plt.show()

# ==========================
# 📉 6. Regressão Linear para Satisfação
# ==========================
print("\n## 📉 Regressão Linear para prever Satisfacao_Curso")
target = 'Satisfacao_Curso'
if target in df.columns:
    X = df_perguntas.copy()
    y = df[target]
    if not X.empty:
        model = LinearRegression()
        model.fit(X, y)
        y_pred = model.predict(X)
        print(f"R²: {r2_score(y, y_pred):.3f}")
        print(f"Intercepto: {model.intercept_:.3f}")
        coef_table = pd.DataFrame({'Variável': X.columns, 'Coeficiente': model.coef_})
        print(coef_table)
        plt.figure(figsize=(8,6))
        plt.scatter(y, y_pred)
        plt.xlabel('Satisfação Real')
        plt.ylabel('Satisfação Predita')
        plt.title('Satisfação Real vs Predita')
        plt.plot([y.min(), y.max()], [y.min(), y.max()], 'k--')
        plt.show()

# ==========================
# 📌 7. Regressão Linear por Pergunta
# ==========================
print("\n## 📌 Regressão Linear - Cada pergunta como dependente")
for target in df_perguntas.columns:
    print(f"\n--- 🔍 Regressão para: {target} ---")
    X = df_perguntas.drop(columns=[target])
    y = df_perguntas[target]
    if X.shape[1] > 0:
        data = pd.concat([X, y], axis=1).dropna()
        X_clean = data[X.columns]
        y_clean = data[target]
        model = LinearRegression()
        model.fit(X_clean, y_clean)
        y_pred = model.predict(X_clean)
        print(f"R²: {r2_score(y_clean, y_pred):.3f}")
        print(f"Intercepto: {model.intercept_:.3f}")
        coef_table = pd.DataFrame({'Variável': X_clean.columns, 'Coeficiente': model.coef_})
        print(coef_table)
        plt.figure(figsize=(7,5))
        plt.scatter(y_clean, y_pred)
        plt.xlabel(f'{target} - Real')
        plt.ylabel(f'{target} - Predito')
        plt.title(f'Regressão Linear: {target}')
        plt.plot([y_clean.min(), y_clean.max()], [y_clean.min(), y_clean.max()], 'k--')
        plt.tight_layout()
        plt.show()

# ==========================
# 🔁 8. Regressão Linear Geral (Todas as Respostas)
# ==========================
print("\n## 📉 Regressão Linear Geral (Todas as respostas combinadas)")
y_all = df_perguntas.values.flatten()
X_all = np.arange(len(y_all)).reshape(-1, 1)
model_all = LinearRegression()
model_all.fit(X_all, y_all)
y_all_pred = model_all.predict(X_all)
plt.figure(figsize=(12, 6))
plt.scatter(X_all, y_all, alpha=0.4, label='Respostas Observadas')
plt.plot(X_all, y_all_pred, color='red', linewidth=2, label='Linha de Regressão')
plt.title('Tendência Geral das Respostas (Regressão Linear)')
plt.xlabel('Índice Sequencial das Respostas')
plt.ylabel('Valor da Resposta')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
print(f"R² da tendência geral: {r2_score(y_all, y_all_pred):.3f}")
print(f"Inclinação da reta: {model_all.coef_[0]:.5f}")
print(f"Intercepto da reta: {model_all.intercept_:.5f}")
```

```python
#!/usr/bin/env python3
"""
echo_pipeline_colab.py

Comprehensive ECHO 1.0 data analysis pipeline adapted for Colab.
Loads a single Excel file via INPUT_PATH, cleans, analyzes, models,
visualizes, and exports results following high cohesion/low coupling,
PEP 8 and PEP 257 guidelines.
"""

import os
from typing import Any, Dict, List, Tuple

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import r2_score

# ─── CONFIGURATION ─────────────────────────────────────────────────────────────

# Path to the uploaded file in Colab
INPUT_PATH = "/content/sample_echo.xlsx"

# Where to save outputs (will be created if needed)
OUTPUT_DIR = "/content/echo_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)


# ─── DATA CLEANING COMPONENTS ──────────────────────────────────────────────────

class DataCleaner:
    """Drops metadata columns and fully empty rows/columns."""

    def __init__(self, metadata_cols: int = 4):
        self.metadata_cols = metadata_cols

    def drop_metadata(self, df: pd.DataFrame) -> pd.DataFrame:
        """Remove the first N metadata columns."""
        return df.iloc[:, self.metadata_cols :]

    def drop_empty(self, df: pd.DataFrame) -> pd.DataFrame:
        """Remove columns and rows that are entirely empty."""
        df = df.dropna(axis=1, how="all")
        return df.dropna(axis=0, how="all")


class DigitExtractor:
    """Extracts the first numeric substring from text columns."""

    @staticmethod
    def extract(df: pd.DataFrame) -> pd.DataFrame:
        for col in df.select_dtypes(include="object"):
            df[col] = (
                df[col].astype(str)
                .str.extract(r"(\d+)", expand=False)
                .fillna(df[col])
            )
        return df


class TypeConverter:
    """Converts columns to numeric where possible."""

    @staticmethod
    def convert(df: pd.DataFrame) -> pd.DataFrame:
        for col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="ignore")
        return df


# ─── ANALYSIS COMPONENTS ───────────────────────────────────────────────────────

class FrequencyCalculator:
    """Computes absolute and relative frequencies."""

    @staticmethod
    def compute(df: pd.DataFrame) -> pd.DataFrame:
        abs_freq = df.apply(lambda s: s.value_counts()).fillna(0).astype(int)
        rel_freq = abs_freq.div(len(df)) * 100
        freq = abs_freq.stack().rename("abs").reset_index()
        freq["rel"] = rel_freq.stack().values
        freq.columns = ["column", "value", "abs", "rel"]
        return freq


class DescriptiveStatsCalculator:
    """Calculates descriptive statistics for numeric data."""

    @staticmethod
    def compute(df: pd.DataFrame) -> pd.DataFrame:
        desc = pd.DataFrame(index=df.columns)
        desc["mean"] = df.mean()
        desc["median"] = df.median()
        desc["std"] = df.std()
        desc["mode"] = df.mode().iloc[0]
        desc["q1"] = df.quantile(0.25)
        desc["q3"] = df.quantile(0.75)
        return desc


class StatisticalTester:
    """Performs chi-square, KS, and Kruskal–Wallis tests."""

    @staticmethod
    def chi2(df: pd.DataFrame) -> Dict[Tuple[str, str], Tuple[float, float]]:
        results: Dict[Tuple[str, str], Tuple[float, float]] = {}
        cat_cols = df.select_dtypes(include="category").columns
        for i, a in enumerate(cat_cols):
            for b in cat_cols[i + 1 :]:
                table = pd.crosstab(df[a], df[b])
                stat, p, _, _ = stats.chi2_contingency(table)
                results[(a, b)] = (stat, p)
        return results

    @staticmethod
    def ks(df: pd.DataFrame) -> Dict[str, Tuple[float, float]]:
        results: Dict[str, Tuple[float, float]] = {}
        for col in df.select_dtypes(include=np.number).columns:
            stat, p = stats.kstest(df[col].dropna(), "norm")
            results[col] = (stat, p)
        return results

    @staticmethod
    def kruskal(df: pd.DataFrame, group_col: str) -> Dict[str, Tuple[float, float]]:
        results: Dict[str, Tuple[float, float]] = {}
        groups = df[group_col].dropna().unique()
        for col in df.select_dtypes(include=np.number).columns:
            data = [df[df[group_col] == g][col].dropna() for g in groups]
            if len(data) > 1:
                stat, p = stats.kruskal(*data)
                results[col] = (stat, p)
        return results


class CorrelationCalculator:
    """Computes Pearson and Spearman correlation matrices."""

    @staticmethod
    def pearson(df: pd.DataFrame) -> pd.DataFrame:
        return df.select_dtypes(include=np.number).corr(method="pearson")

    @staticmethod
    def spearman(df: pd.DataFrame) -> pd.DataFrame:
        return df.select_dtypes(include=np.number).corr(method="spearman")


class ReliabilityCalculator:
    """Computes Cronbach's alpha for scale reliability."""

    @staticmethod
    def cronbach_alpha(df: pd.DataFrame) -> float:
        items = df.to_numpy()
        item_var = items.var(axis=0, ddof=1)
        total_var = items.sum(axis=1).var(ddof=1)
        n = items.shape[1]
        return (n / (n - 1)) * (1 - item_var.sum() / total_var)


class RegressionModeler:
    """Fits linear and Ridge regression models."""

    @staticmethod
    def fit(
        df: pd.DataFrame, target: str, features: List[str]
    ) -> Dict[str, Dict[str, Any]]:
        X = df[features].dropna()
        y = df.loc[X.index, target]
        lm = LinearRegression().fit(X, y)
        ridge = Ridge().fit(X, y)
        return {
            "linear": {"model": lm, "r2": r2_score(y, lm.predict(X))},
            "ridge": {"model": ridge, "r2": r2_score(y, ridge.predict(X))},
        }


# ─── VISUALIZATION COMPONENTS ─────────────────────────────────────────────────

class Visualizer:
    """Generates and saves required plots."""

    @staticmethod
    def histogram(df: pd.DataFrame, output_dir: str) -> None:
        for col in df.select_dtypes(include=np.number).columns:
            plt.figure()
            df[col].hist(bins=20)
            plt.title(f"Histogram of {col}")
            plt.savefig(os.path.join(output_dir, f"hist_{col}.png"))
            plt.close()

    @staticmethod
    def boxplot(df: pd.DataFrame, output_dir: str) -> None:
        for col in df.select_dtypes(include=np.number).columns:
            plt.figure()
            df.boxplot(column=col)
            plt.title(f"Boxplot of {col}")
            plt.savefig(os.path.join(output_dir, f"box_{col}.png"))
            plt.close()

    @staticmethod
    def heatmap(corr: pd.DataFrame, output_path: str) -> None:
        plt.figure(figsize=(8, 6))
        plt.imshow(corr, interpolation="nearest")
        plt.colorbar()
        plt.xticks(range(len(corr)), corr.columns, rotation=90)
        plt.yticks(range(len(corr)), corr.columns)
        plt.tight_layout()
        plt.savefig(output_path)
        plt.close()

    @staticmethod
    def trend_over_time(
        df: pd.DataFrame, sort_col: str, output_path: str
    ) -> None:
        ser = df[sort_col].dropna().reset_index(drop=True)
        x = np.arange(len(ser)).reshape(-1, 1)
        y = ser.values
        model = LinearRegression().fit(x, y)
        trend = model.predict(x)

        plt.figure()
        plt.plot(x, y, marker="o", linestyle="-", label=sort_col)
        plt.plot(x, trend, linestyle="--", label="Trend")
        plt.title(f"Sequential plot and trend for {sort_col}")
        plt.legend()
        plt.savefig(output_path)
        plt.close()


class ResultsExporter:
    """Writes all results to CSV or text files."""

    @staticmethod
    def export_all(
        freq: pd.DataFrame,
        desc: pd.DataFrame,
        chi2_res: Dict,
        ks_res: Dict,
        kruskal_res: Dict,
        pearson_corr: pd.DataFrame,
        spearman_corr: pd.DataFrame,
        alpha: float,
        regs: Dict,
        output_dir: str,
    ) -> None:
        freq.to_csv(os.path.join(output_dir, "frequencies.csv"), index=False)
        desc.to_csv(os.path.join(output_dir, "descriptive_stats.csv"))

        pd.DataFrame(
            [{"pair": f"{a} vs {b}", "chi2": s, "p": p}
             for (a, b), (s, p) in chi2_res.items()]
        ).to_csv(os.path.join(output_dir, "chi2_tests.csv"), index=False)

        pd.DataFrame(
            [{"column": c, "ks_stat": s, "p": p}
             for c, (s, p) in ks_res.items()]
        ).to_csv(os.path.join(output_dir, "ks_tests.csv"), index=False)

        pd.DataFrame(
            [{"variable": v, "kw_stat": s, "p": p}
             for v, (s, p) in kruskal_res.items()]
        ).to_csv(os.path.join(output_dir, "kruskal_tests.csv"), index=False)

        pearson_corr.to_csv(os.path.join(output_dir, "pearson_corr.csv"))
        spearman_corr.to_csv(os.path.join(output_dir, "spearman_corr.csv"))

        with open(os.path.join(output_dir, "cronbach_alpha.txt"), "w") as f:
            f.write(str(alpha))

        pd.DataFrame(
            [{name: info["r2"] for name, info in regs.items()}],
            index=["r2_score"],
        ).T.to_csv(os.path.join(output_dir, "regression_r2_scores.csv"))


# ─── PIPELINE ORCHESTRATOR ────────────────────────────────────────────────────

# … importações e classes …

class EchoPipeline:
    """Orchestrates the full end-to-end analysis."""

    def __init__(
        self,
        input_path: str,
        group_col: str,
        trend_col: str,
    ):
        self.input_path = input_path
        self.group_col = group_col
        self.trend_col = trend_col
        self.cleaner = DataCleaner()

    def run(self) -> None:
        """
        Load the data, clean, analyze, visualize, and export results.
        """
        # 1. Load, escolhendo engine conforme a extensão
        ext = os.path.splitext(self.input_path)[1].lower()
        if ext == ".csv":
            df = pd.read_csv(self.input_path)
        elif ext in (".xls", ".xlsx"):
            df = pd.read_excel(self.input_path, engine="openpyxl")
        else:
            raise ValueError(f"Unsupported file type: {ext}")

        df[self.group_col] = df[self.group_col].astype("category")

        # 2. Clean
        df = self.cleaner.drop_empty(
            DigitExtractor.extract(
                TypeConverter.convert(self.cleaner.drop_metadata(df))
            )
        )



def main() -> None:
    """Entrypoint when running as a script."""
    pipeline = EchoPipeline(
        input_path=INPUT_PATH,
        group_col="GroupColumn",
        trend_col="YourTrendColumn",
    )
    pipeline.run()


def main() -> None:
    """Entrypoint when running as a script."""
    pipeline = EchoPipeline(
        input_path=INPUT_PATH,
        group_col="GroupColumn",
        trend_col="YourTrendColumn",
    )
    pipeline.run()


if __name__ == "__main__":
    main()
```
